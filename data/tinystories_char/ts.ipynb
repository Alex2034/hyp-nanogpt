```python
import os
import numpy as np
from datasets import load_dataset
    
train_dataset = load_dataset("roneneldan/TinyStories", split="train")
train_texts = ''.join(train_dataset["text"])

print(f"Length of train text: {len(train_texts)} characters")
```

```python
import matplotlib.pyplot as plt
from collections import Counter
text = train_texts[:100_000_000]
counts = Counter(text)
chars = sorted(counts.keys())
freqs = sorted(counts.values(), reverse=True)
```

```python
freqs = sorted(counts.values(), reverse=True)
plt.figure()
plt.loglog(freqs, color='orange')
plt.ylabel('Count')
plt.title('Character Frequency Distribution')
plt.show()
```

```python
counts
```

```python
allowed = {c for c,f in counts.items() if f > 10}
```

```python
filtered = ''.join(c for c in text if c in allowed)

with open('input.txt', 'w', encoding='utf-8') as f:
    f.write(filtered)
```
