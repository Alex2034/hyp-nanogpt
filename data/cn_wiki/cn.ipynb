```python
from datasets import load_dataset

ds = load_dataset("pleisto/wikipedia-cn-20230720-filtered")
```

```python
text = ''.join(ds['train']['completion'])
```

```python
import matplotlib.pyplot as plt
from collections import Counter

counts = Counter(text)
chars = sorted(counts.keys())
```

```python
freqs = sorted(counts.values(), reverse=True)
plt.figure()
plt.loglog(freqs, color='orange')
plt.loglog([1000], [freqs[1000]], markersize=7, marker='o', color='r')
plt.ylabel('Count')
plt.title('Character Frequency Distribution')
plt.show()
```

```python
allowed = {c for c,f in counts.items() if f > 100}
```

```python
len(allowed)
```

```python
filtered = ''.join(c for c in text if c in allowed)

with open('input.txt', 'w', encoding='utf-8') as f:
    f.write(filtered)
```

```python
counts_f = Counter(filtered)
freqs_f = sorted(counts_f.values(), reverse=True)
```
