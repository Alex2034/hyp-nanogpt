```python
from datasets import load_dataset

ds = load_dataset("pleisto/wikipedia-cn-20230720-filtered")
```

```python
text = ''.join(ds['train']['completion'])
```

```python
import matplotlib.pyplot as plt
from collections import Counter

counts = Counter(text)
chars = sorted(counts.keys())
```

```python
freqs = sorted(counts.values(), reverse=True)
plt.figure()
plt.plot(freqs, color='crimson', lw=2)
plt.plot([6421], [freqs[6421]], markersize=7, marker='o', color='darkred', label='Truncation')
plt.plot([0, 16421], [freqs[6421], freqs[6421]], 'k--', alpha=0.2)
plt.ylabel('Count')
plt.yscale('log')
plt.xscale('log')
plt.legend()
plt.title('Token Frequency Distribution')
plt.show()
```

```python
allowed = {c for c,f in counts.items() if f > 100}
```
